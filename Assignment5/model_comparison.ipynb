```python
# ============================================================
# DEEP LEARNING PRACTICAL
# RNN vs LSTM vs GRU for Sequence Classification
# Dataset: IMDB Movie Reviews
# ============================================================

import tensorflow as tf
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import time

from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing.sequence import pad_sequences

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import (
    Embedding,
    SimpleRNN,
    LSTM,
    GRU,
    Dense,
    Dropout
)

from tensorflow.keras.optimizers import Adam

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report,
    roc_curve,
    auc
)


# ============================================================
# 1. REPRODUCIBILITY
# ============================================================

tf.random.set_seed(42)
np.random.seed(42)


# ============================================================
# 2. HARD-CODED EXPERIMENT SETTINGS
# ============================================================

VOCAB_SIZE = 10000
MAX_LENGTH = 200
EMBEDDING_DIM = 64
RNN_UNITS = 64

EPOCHS = 5
BATCH_SIZE = 128
LEARNING_RATE = 0.001


print("=" * 70)
print("       RNN vs LSTM vs GRU - SEQUENCE CLASSIFICATION")
print("=" * 70)

print("\nExperiment Configuration")
print("-" * 70)
print("Vocabulary Size       :", VOCAB_SIZE)
print("Maximum Sequence Len  :", MAX_LENGTH)
print("Embedding Dimension   :", EMBEDDING_DIM)
print("Recurrent Units       :", RNN_UNITS)
print("Epochs                :", EPOCHS)
print("Batch Size            :", BATCH_SIZE)
print("Learning Rate         :", LEARNING_RATE)


# ============================================================
# 3. LOAD IMDB DATASET
# ============================================================

print("\n" + "=" * 70)
print("LOADING IMDB DATASET")
print("=" * 70)

print("\nLoading IMDB dataset using Keras...")

(x_train, y_train), (x_test, y_test) = imdb.load_data(
    num_words=VOCAB_SIZE
)

print("\nDataset loaded successfully.")

print("Training samples :", len(x_train))
print("Testing samples  :", len(x_test))

print("\nExample sequence:")
print(x_train[0][:20])

print("\nExample label:")
print(
    "Positive"
    if y_train[0] == 1
    else "Negative"
)


# ============================================================
# 4. SEQUENCE PREPROCESSING
# ============================================================

print("\n" + "=" * 70)
print("SEQUENCE PREPROCESSING")
print("=" * 70)

print("\nOriginal sequence length:")
print(len(x_train[0]))

x_train = pad_sequences(
    x_train,
    maxlen=MAX_LENGTH,
    padding="post",
    truncating="post"
)

x_test = pad_sequences(
    x_test,
    maxlen=MAX_LENGTH,
    padding="post",
    truncating="post"
)

print("\nAfter padding/truncation:")

print("Training shape :", x_train.shape)
print("Testing shape  :", x_test.shape)

print(
    "\nEvery review is represented using",
    MAX_LENGTH,
    "tokens."
)


# ============================================================
# 5. MODEL CREATION FUNCTION
# ============================================================

def create_model(model_type):

    model = Sequential()

    model.add(
        Embedding(
            input_dim=VOCAB_SIZE,
            output_dim=EMBEDDING_DIM
        )
    )

    # --------------------------------------------------------
    # Recurrent Layer
    # --------------------------------------------------------

    if model_type == "RNN":

        model.add(
            SimpleRNN(
                RNN_UNITS
            )
        )

    elif model_type == "LSTM":

        model.add(
            LSTM(
                RNN_UNITS
            )
        )

    elif model_type == "GRU":

        model.add(
            GRU(
                RNN_UNITS
            )
        )

    # --------------------------------------------------------
    # Classification Layers
    # --------------------------------------------------------

    model.add(
        Dropout(0.3)
    )

    model.add(
        Dense(
            32,
            activation="relu"
        )
    )

    model.add(
        Dense(
            1,
            activation="sigmoid"
        )
    )

    # --------------------------------------------------------
    # Compile
    # --------------------------------------------------------

    model.compile(
        optimizer=Adam(
            learning_rate=LEARNING_RATE
        ),
        loss="binary_crossentropy",
        metrics=["accuracy"]
    )

    return model


# ============================================================
# 6. MODEL ARCHITECTURE
# ============================================================

print("\n" + "=" * 70)
print("MODEL ARCHITECTURE")
print("=" * 70)

print("""
                    IMDB REVIEW
                         |
                         v
                Integer Token Sequence
                         |
                         v
                    EMBEDDING
                         |
          +--------------+--------------+
          |              |              |
          v              v              v
        RNN            LSTM            GRU
          |              |              |
          +--------------+--------------+
                         |
                       Dropout
                         |
                     Dense (32)
                         |
                  Sigmoid Output
                         |
                  Positive / Negative
""")


# ============================================================
# 7. TRAIN RNN, LSTM AND GRU
# ============================================================

model_types = [
    "RNN",
    "LSTM",
    "GRU"
]

models = {}
histories = {}
predictions = {}
results = {}
training_times = {}


for model_type in model_types:

    print("\n")
    print("=" * 70)
    print(f"TRAINING {model_type}")
    print("=" * 70)

    model = create_model(model_type)

    print("\nModel Summary:")
    model.summary()

    start_time = time.time()

    history = model.fit(
        x_train,
        y_train,
        validation_split=0.2,
        epochs=EPOCHS,
        batch_size=BATCH_SIZE,
        verbose=1
    )

    end_time = time.time()

    training_time = end_time - start_time

    models[model_type] = model
    histories[model_type] = history
    training_times[model_type] = training_time

    print(
        f"\n{model_type} training completed "
        f"in {training_time:.2f} seconds."
    )


# ============================================================
# 8. MODEL EVALUATION
# ============================================================

print("\n" + "=" * 70)
print("MODEL EVALUATION")
print("=" * 70)


for model_type in model_types:

    model = models[model_type]

    test_loss, test_accuracy = model.evaluate(
        x_test,
        y_test,
        verbose=0
    )

    y_probability = model.predict(
        x_test,
        verbose=0
    ).ravel()

    y_prediction = (
        y_probability >= 0.5
    ).astype(int)

    predictions[model_type] = y_probability

    accuracy = accuracy_score(
        y_test,
        y_prediction
    )

    precision = precision_score(
        y_test,
        y_prediction
    )

    recall = recall_score(
        y_test,
        y_prediction
    )

    f1 = f1_score(
        y_test,
        y_prediction
    )

    results[model_type] = {
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1 Score": f1,
        "Test Loss": test_loss,
        "Parameters": model.count_params(),
        "Training Time": training_times[model_type]
    }


# ============================================================
# 9. FINAL COMPARISON TABLE
# ============================================================

comparison_df = pd.DataFrame(
    results
).T

comparison_df["Accuracy"] = \
    comparison_df["Accuracy"].round(4)

comparison_df["Precision"] = \
    comparison_df["Precision"].round(4)

comparison_df["Recall"] = \
    comparison_df["Recall"].round(4)

comparison_df["F1 Score"] = \
    comparison_df["F1 Score"].round(4)

comparison_df["Test Loss"] = \
    comparison_df["Test Loss"].round(4)

comparison_df["Training Time"] = \
    comparison_df["Training Time"].round(2)


print("\n")
print("=" * 100)
print("FINAL COMPARISON TABLE")
print("=" * 100)

print(comparison_df)


# ============================================================
# 10. METRIC-WISE RESULTS
# ============================================================

print("\n")
print("=" * 70)
print("METRIC-WISE COMPARISON")
print("=" * 70)

print(
    "\nHighest Accuracy :",
    comparison_df["Accuracy"].idxmax()
)

print(
    "Highest Precision:",
    comparison_df["Precision"].idxmax()
)

print(
    "Highest Recall   :",
    comparison_df["Recall"].idxmax()
)

print(
    "Highest F1 Score :",
    comparison_df["F1 Score"].idxmax()
)

print(
    "Lowest Test Loss :",
    comparison_df["Test Loss"].idxmin()
)


# ============================================================
# 11. TRAINING ACCURACY
# ============================================================

plt.figure(figsize=(10, 6))

for model_type in model_types:

    plt.plot(
        histories[model_type].history["accuracy"],
        marker="o",
        label=model_type
    )

plt.title(
    "Training Accuracy - RNN vs LSTM vs GRU"
)

plt.xlabel("Epoch")
plt.ylabel("Training Accuracy")

plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()


# ============================================================
# 12. VALIDATION ACCURACY
# ============================================================

plt.figure(figsize=(10, 6))

for model_type in model_types:

    plt.plot(
        histories[model_type].history["val_accuracy"],
        marker="o",
        label=model_type
    )

plt.title(
    "Validation Accuracy - RNN vs LSTM vs GRU"
)

plt.xlabel("Epoch")
plt.ylabel("Validation Accuracy")

plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()


# ============================================================
# 13. TRAINING LOSS
# ============================================================

plt.figure(figsize=(10, 6))

for model_type in model_types:

    plt.plot(
        histories[model_type].history["loss"],
        marker="o",
        label=model_type
    )

plt.title(
    "Training Loss - RNN vs LSTM vs GRU"
)

plt.xlabel("Epoch")
plt.ylabel("Training Loss")

plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()


# ============================================================
# 14. VALIDATION LOSS
# ============================================================

plt.figure(figsize=(10, 6))

for model_type in model_types:

    plt.plot(
        histories[model_type].history["val_loss"],
        marker="o",
        label=model_type
    )

plt.title(
    "Validation Loss - RNN vs LSTM vs GRU"
)

plt.xlabel("Epoch")
plt.ylabel("Validation Loss")

plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()


# ============================================================
# 15. ACCURACY / PRECISION / RECALL / F1 COMPARISON
# ============================================================

metrics = [
    "Accuracy",
    "Precision",
    "Recall",
    "F1 Score"
]

comparison_df[metrics].plot(
    kind="bar",
    figsize=(11, 6)
)

plt.title(
    "RNN vs LSTM vs GRU - Performance Metrics"
)

plt.xlabel("Model")
plt.ylabel("Score")

plt.ylim(0, 1)

plt.xticks(rotation=0)

plt.grid(axis="y")

plt.tight_layout()
plt.show()


# ============================================================
# 16. TEST LOSS COMPARISON
# ============================================================

plt.figure(figsize=(8, 5))

plt.bar(
    comparison_df.index,
    comparison_df["Test Loss"]
)

plt.title(
    "Test Loss Comparison"
)

plt.xlabel("Model")
plt.ylabel("Test Loss")

plt.grid(axis="y")

plt.tight_layout()
plt.show()


# ============================================================
# 17. PARAMETER COUNT
# ============================================================

plt.figure(figsize=(8, 5))

plt.bar(
    comparison_df.index,
    comparison_df["Parameters"]
)

plt.title(
    "Trainable Parameter Comparison"
)

plt.xlabel("Model")
plt.ylabel("Number of Parameters")

plt.grid(axis="y")

plt.tight_layout()
plt.show()


# ============================================================
# 18. TRAINING TIME
# ============================================================

plt.figure(figsize=(8, 5))

plt.bar(
    comparison_df.index,
    comparison_df["Training Time"]
)

plt.title(
    "Training Time Comparison"
)

plt.xlabel("Model")
plt.ylabel("Training Time (seconds)")

plt.grid(axis="y")

plt.tight_layout()
plt.show()


# ============================================================
# 19. CONFUSION MATRICES
# ============================================================

fig, axes = plt.subplots(
    1,
    3,
    figsize=(15, 4)
)


for i, model_type in enumerate(model_types):

    y_prediction = (
        predictions[model_type] >= 0.5
    ).astype(int)

    cm = confusion_matrix(
        y_test,
        y_prediction
    )

    sns.heatmap(
        cm,
        annot=True,
        fmt="d",
        cmap="Blues",
        ax=axes[i],
        xticklabels=[
            "Negative",
            "Positive"
        ],
        yticklabels=[
            "Negative",
            "Positive"
        ]
    )

    axes[i].set_title(
        f"{model_type} Confusion Matrix"
    )

    axes[i].set_xlabel("Predicted")
    axes[i].set_ylabel("Actual")


plt.tight_layout()
plt.show()


# ============================================================
# 20. ROC CURVES
# ============================================================

plt.figure(figsize=(9, 6))

for model_type in model_types:

    y_probability = predictions[
        model_type
    ]

    fpr, tpr, _ = roc_curve(
        y_test,
        y_probability
    )

    roc_auc = auc(
        fpr,
        tpr
    )

    plt.plot(
        fpr,
        tpr,
        label=f"{model_type} (AUC = {roc_auc:.3f})"
    )


plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--"
)

plt.title(
    "ROC Curve - RNN vs LSTM vs GRU"
)

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")

plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()


# ============================================================
# 21. CLASSIFICATION REPORTS
# ============================================================

print("\n")
print("=" * 70)
print("DETAILED CLASSIFICATION REPORTS")
print("=" * 70)


for model_type in model_types:

    y_prediction = (
        predictions[model_type] >= 0.5
    ).astype(int)

    print("\n")
    print("-" * 70)
    print(model_type)
    print("-" * 70)

    print(
        classification_report(
            y_test,
            y_prediction,
            target_names=[
                "Negative",
                "Positive"
            ]
        )
    )


# ============================================================
# 22. SAMPLE PREDICTIONS
# ============================================================

print("\n")
print("=" * 70)
print("SAMPLE PREDICTIONS")
print("=" * 70)


sample_indices = [
    10,
    50,
    100,
    500,
    1000
]

sample_results = []


for index in sample_indices:

    row = {
        "Sample": index,
        "Actual": (
            "Positive"
            if y_test[index] == 1
            else "Negative"
        )
    }

    for model_type in model_types:

        probability = float(
            predictions[model_type][index]
        )

        row[
            f"{model_type} Prediction"
        ] = (
            "Positive"
            if probability >= 0.5
            else "Negative"
        )

        row[
            f"{model_type} Probability"
        ] = round(
            probability,
            3
        )

    sample_results.append(row)


sample_df = pd.DataFrame(
    sample_results
)

print(
    sample_df.to_string(
        index=False
    )
)


# ============================================================
# 24. FINAL RESULT
# ============================================================

print("\n")
print("=" * 70)
print("FINAL RESULT")
print("=" * 70)

print(comparison_df)

print("\nRNN vs LSTM vs GRU comparison completed successfully.")
```
